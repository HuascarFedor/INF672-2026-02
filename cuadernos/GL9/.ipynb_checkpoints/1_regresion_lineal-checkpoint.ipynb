```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Paleta del curso (cuaderno científico)
INDIGO, PURPURA, CORAL, TINTA = '#4F46E5', '#7E22CE', '#F43F5E', '#111827'
plt.rcParams.update({'figure.figsize': (7, 4.2), 'axes.grid': True,
                     'grid.alpha': 0.3, 'axes.spines.top': False,
                     'axes.spines.right': False})
np.set_printoptions(precision=4, suppress=True)
```

## 1. El conjunto de datos

Trabajaremos primero con un conjunto **mínimo** de 5 observaciones. Es tan pequeño que podemos verificar cada cálculo a mano, lo que lo hace ideal para comprender el algoritmo antes de pasar a datos más grandes.

Imaginemos que $x$ son las **horas de estudio** de un estudiante y $y$ la **nota** obtenida en una práctica (escala 0–5 para simplificar).

```python
X = np.array([[1], [2], [3], [4], [5]])   # matriz de características: forma (m, n) = (5, 1)
y = np.array([2, 4, 5, 4, 5])             # vector objetivo: forma (m,) = (5,)

print('Forma de X:', X.shape)
print('Forma de y:', y.shape)
```

<div style="border-left:4px solid #6B7280; background:#F9FAFB; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#6B7280; font-weight:bold;">NOTA</span> <b>¿Por qué X tiene doble corchete?</b><br>scikit-learn espera que <code>X</code> sea una <b>matriz 2D</b> de forma <code>(m muestras, n características)</code>, aunque solo haya una característica. Por eso escribimos <code>[[1],[2],...]</code>. El vector objetivo <code>y</code>, en cambio, es 1D.</div>

```python
plt.scatter(X, y, s=70, color=INDIGO, zorder=3)
plt.xlabel('x  (horas de estudio)')
plt.ylabel('y  (nota)')
plt.title('Datos de entrenamiento')
plt.xlim(0, 6); plt.ylim(0, 6)
plt.show()
```

Los puntos no están perfectamente alineados: **no existe una recta que pase por todos**. El objetivo será encontrar la recta que, en conjunto, *mejor* los aproxime. Para eso necesitamos definir qué significa "mejor".

## 2. Modelo y función de costo

### 2.1 Hipótesis

El modelo de regresión lineal simple propone:

$$\hat{y} = f_{w,b}(x) = w\,x + b$$

- $w$ es la **pendiente** (peso): cuánto cambia la predicción por cada unidad de $x$.
- $b$ es el **intercepto** (sesgo): el valor predicho cuando $x = 0$.

### 2.2 Error cuadrático medio

Para cada observación $i$, el **residuo** es la diferencia entre lo predicho y lo real: $e^{(i)} = \hat{y}^{(i)} - y^{(i)}$. Medimos la calidad de un par $(w, b)$ con la función de costo:

$$J(w,b) = \frac{1}{2m}\sum_{i=1}^{m}\left(w\,x^{(i)} + b - y^{(i)}\right)^2$$

<div style="border-left:4px solid #7E22CE; background:#F5F3FF; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#7E22CE; font-weight:bold;">DEFINICIÓN</span> <b>Función de costo J(w, b)</b><br>Es un número que resume <i>cuánto se equivoca</i> el modelo sobre todo el conjunto de entrenamiento. Entrenar el modelo es <b>encontrar el par (w, b) que minimiza J</b>. Elevamos al cuadrado para que los errores positivos y negativos no se cancelen y para penalizar más los errores grandes. El factor ½ es una convención: simplifica la derivada (el 2 del exponente se cancela) y no cambia dónde está el mínimo.</div>

```python
def costo(w, b, x, y):
    """Error cuadrático medio con factor 1/2:  J(w,b) = (1/2m) Σ (w·x + b − y)²."""
    m = len(y)
    residuos = w * x + b - y
    return (residuos ** 2).sum() / (2 * m)

x = X.ravel()   # versión 1D de X para las operaciones elemento a elemento

# Probemos algunas rectas "a ojo"
for w_, b_ in [(0, 0), (1, 1), (0.5, 2.5), (0.6, 2.2)]:
    print(f'w = {w_:<4}  b = {b_:<4}  ->  J = {costo(w_, b_, x, y):.4f}')
```

Algunas rectas son claramente peores que otras. La última, $(0.6,\ 2.2)$, parece dar el menor costo: veremos enseguida que **es exactamente el mínimo**.

Visualicemos $J$ como una superficie: cada punto del plano $(w, b)$ tiene un costo asociado. Las curvas de nivel muestran que $J$ tiene forma de **cuenco** (es una función convexa) con un único mínimo.

```python
W, B = np.meshgrid(np.linspace(-1, 2.2, 200), np.linspace(-1, 5, 200))
J = ((W[..., None] * x + B[..., None] - y) ** 2).mean(axis=-1) / 2

plt.contourf(W, B, J, levels=30, cmap='Purples')
plt.colorbar(label='J(w, b)')
plt.contour(W, B, J, levels=[0.3, 0.5, 1, 2, 4, 8], colors='white', linewidths=0.6)
plt.scatter(0.6, 2.2, color=CORAL, s=80, marker='*', zorder=3, label='mínimo (0.6, 2.2)')
plt.xlabel('w'); plt.ylabel('b'); plt.title('Curvas de nivel de la función de costo')
plt.legend(); plt.show()
```

<div style="border-left:4px solid #4F46E5; background:#EEF2FF; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#4F46E5; font-weight:bold;">IDEA CLAVE</span> <b>Convexidad</b><br>Como <i>J</i> es convexa, <b>cualquier mínimo local es también el global</b>. Esto garantiza que el descenso de gradiente, con una tasa de aprendizaje adecuada, llegará a la misma solución que las fórmulas cerradas. En redes neuronales la superficie deja de ser convexa y esta garantía se pierde.</div>

## 3. Camino A — Solución cerrada (mínimos cuadrados)

Si igualamos a cero las derivadas parciales de $J$ y despejamos, obtenemos fórmulas directas:

$$w = \frac{\sum_{i}(x^{(i)}-\bar{x})(y^{(i)}-\bar{y})}{\sum_{i}(x^{(i)}-\bar{x})^2} \qquad\qquad b = \bar{y} - w\,\bar{x}$$

donde $\bar{x}$ y $\bar{y}$ son las medias. El numerador mide cómo **covarían** $x$ e $y$; el denominador, cuánto **varía** $x$.

### 3.1 Cálculo paso a paso

```python
x_media, y_media = x.mean(), y.mean()
dx, dy = x - x_media, y - y_media

print(f'x̄ = {x_media},  ȳ = {y_media}')
print('x − x̄          :', dx)
print('y − ȳ          :', dy)
print('(x − x̄)(y − ȳ) :', dx * dy, ' suma =', (dx * dy).sum())
print('(x − x̄)²       :', dx ** 2, ' suma =', (dx ** 2).sum())

w_cerrada = (dx * dy).sum() / (dx ** 2).sum()
b_cerrada = y_media - w_cerrada * x_media
print(f'\nw = {(dx*dy).sum()} / {(dx**2).sum()} = {w_cerrada:.3f}')
print(f'b = {y_media} − {w_cerrada:.1f}·{x_media} = {b_cerrada:.3f}')
```

### 3.2 Ecuación normal (forma matricial)

La misma solución se expresa matricialmente. Agregamos una columna de unos a $X$ para que el intercepto $b$ se trate como un peso más:

$$\tilde{X} = \begin{bmatrix} 1 & x^{(1)} \\ \vdots & \vdots \\ 1 & x^{(m)} \end{bmatrix}, \qquad \boldsymbol{\theta} = \begin{bmatrix} b \\ w \end{bmatrix} = \left(\tilde{X}^{\top}\tilde{X}\right)^{-1}\tilde{X}^{\top}\mathbf{y}$$

Esta forma es la que generaliza sin cambios al caso de **muchas variables**.

```python
X_tilde = np.c_[np.ones(len(x)), x]          # añade la columna de unos
print('X̃ =\n', X_tilde)

# Resolvemos (X̃ᵀX̃) θ = X̃ᵀy  (más estable que invertir la matriz explícitamente)
theta = np.linalg.solve(X_tilde.T @ X_tilde, X_tilde.T @ y)
print(f'\nθ = [b, w] = {theta}')
```

<div style="border-left:4px solid #F43F5E; background:#FFF1F2; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#F43F5E; font-weight:bold;">ADVERTENCIA</span> <b>No invertir matrices si se puede evitar</b><br>Escribimos la fórmula con la inversa, pero en código conviene usar <code>np.linalg.solve</code> o <code>np.linalg.lstsq</code>: son numéricamente más estables y más rápidos. Además, invertir <code>X̃ᵀX̃</code> cuesta del orden de <i>n</i>³ operaciones, lo que se vuelve prohibitivo con miles de características — una de las razones para preferir el descenso de gradiente.</div>

## 4. Camino A' — La misma solución con scikit-learn

`LinearRegression` resuelve internamente el problema de mínimos cuadrados (con una descomposición numérica estable). La API de scikit-learn sigue siempre el mismo patrón: **crear → `fit` → `predict`**.

```python
modelo = LinearRegression()   # 1. crear el estimador
modelo.fit(X, y)              # 2. ajustar (entrenar) con los datos

print('w =', round(modelo.coef_[0], 3))       # coef_: arreglo con un peso por característica
print('b =', round(modelo.intercept_, 3))     # intercept_: escalar
```

```python
y_pred = modelo.predict(X)    # 3. predecir

print('y real     :', y)
print('y predicha :', y_pred)
print('residuos   :', y_pred - y)
print(f'\nMSE = {mean_squared_error(y, y_pred):.4f}')
print(f'R²  = {r2_score(y, y_pred):.4f}')
print(f'J   = MSE/2 = {costo(modelo.coef_[0], modelo.intercept_, x, y):.4f}')

# Predicción para un valor nuevo: ¿qué nota se espera con 6 horas?
print(f'\nPredicción para x = 6: {modelo.predict([[6]])[0]:.2f}')
```

<div style="border-left:4px solid #7E22CE; background:#F5F3FF; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#7E22CE; font-weight:bold;">DEFINICIÓN</span> <b>Coeficiente de determinación R²</b><br>Indica qué proporción de la variabilidad de <i>y</i> explica el modelo: 1 es un ajuste perfecto, 0 equivale a predecir siempre la media. Aquí R² = 0.6: la recta explica el 60 % de la variación de las notas. Observa también que <code>mean_squared_error</code> no lleva el factor ½, por eso <i>J</i> = MSE/2.</div>

```python
xs = np.linspace(0, 6, 100)
plt.scatter(X, y, s=70, color=INDIGO, zorder=3, label='datos')
plt.plot(xs, modelo.predict(xs.reshape(-1, 1)), color=PURPURA, lw=2, label='ŷ = 0.6x + 2.2')
for xi, yi, ypi in zip(x, y, y_pred):
    plt.vlines(xi, yi, ypi, color=CORAL, ls='--', lw=1.2)
plt.plot([], [], color=CORAL, ls='--', label='residuos')
plt.xlabel('x'); plt.ylabel('y'); plt.xlim(0, 6); plt.ylim(0, 6)
plt.title('Recta de mínimos cuadrados'); plt.legend(); plt.show()
```

La recta minimiza la **suma de los cuadrados** de los segmentos coral. Cualquier otra recta produciría una suma mayor.

## 5. Camino B — Descenso de gradiente desde cero

La solución cerrada da la respuesta "de golpe". El descenso de gradiente, en cambio, parte de un punto cualquiera y **camina cuesta abajo** sobre la superficie de $J$ hasta llegar al fondo del cuenco.

### 5.1 Las derivadas parciales

Aplicando la regla de la cadena a $J(w,b)$:

$$\frac{\partial J}{\partial w} = \frac{1}{m}\sum_{i=1}^{m}\underbrace{\left(w\,x^{(i)} + b - y^{(i)}\right)}_{e^{(i)}}\,x^{(i)} \qquad\qquad \frac{\partial J}{\partial b} = \frac{1}{m}\sum_{i=1}^{m}e^{(i)}$$

El 2 que baja del exponente se cancela con el ½ de la definición: esa era la razón del factor ½.

### 5.2 La regla de actualización

El gradiente apunta hacia donde $J$ **crece** más rápido; por eso nos movemos en sentido **contrario**, con un tamaño de paso $\alpha$ (tasa de aprendizaje):

$$w \leftarrow w - \alpha\,\frac{\partial J}{\partial w} \qquad\qquad b \leftarrow b - \alpha\,\frac{\partial J}{\partial b}$$

<div style="border-left:4px solid #F43F5E; background:#FFF1F2; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#F43F5E; font-weight:bold;">ADVERTENCIA</span> <b>Actualización simultánea</b><br>Ambos gradientes deben calcularse con los valores <b>anteriores</b> de <i>w</i> y <i>b</i>, y solo después se actualizan los dos parámetros. Si actualizáramos <i>w</i> primero y usáramos el <i>w</i> nuevo para calcular <code>db</code>, estaríamos implementando otro algoritmo. En el código siguiente esto se cumple porque <code>dw</code> y <code>db</code> se calculan antes de modificar nada.</div>

### 5.3 Un paso a mano

Antes de iterar miles de veces, hagamos **una sola iteración** partiendo de $w = b = 0$ con $\alpha = 0.01$, para ver exactamente qué ocurre.

```python
w, b, alpha = 0.0, 0.0, 0.01

y_pred = w * x + b                  # todas las predicciones son 0
error  = y_pred - y                 # e = ŷ − y
dw = (error * x).mean()             # ∂J/∂w
db = error.mean()                   # ∂J/∂b
print('predicciones:', y_pred)
print('errores     :', error)
print(f'dw = {dw:.2f},  db = {db:.2f}')

w, b = w - alpha * dw, b - alpha * db
print(f'Tras 1 paso: w = {w:.4f}, b = {b:.4f},  J = {costo(w, b, x, y):.4f}  (antes J = {costo(0, 0, x, y):.4f})')
```

Los gradientes son negativos (el modelo predice por debajo de los datos), así que restar $\alpha \cdot$gradiente **aumenta** $w$ y $b$, y el costo baja. Ahora automatizamos el proceso y guardamos el historial para poder analizarlo.

### 5.4 Implementación completa

```python
def descenso_gradiente(x, y, alpha=0.01, iteraciones=10_000, w0=0.0, b0=0.0, guardar_cada=1):
    """Regresión lineal simple por descenso de gradiente (batch).

    Devuelve los parámetros finales y el historial de (w, b, J)."""
    w, b = w0, b0
    historial = []
    for it in range(iteraciones):
        # 1) Propagación hacia adelante: predicciones y errores
        y_pred = w * x + b
        error  = y_pred - y
        # 2) Gradientes (con los valores actuales de w y b)
        dw = (error * x).mean()
        db = error.mean()
        # 3) Actualización simultánea
        w -= alpha * dw
        b -= alpha * db
        if it % guardar_cada == 0:
            historial.append((w, b, costo(w, b, x, y)))
    return w, b, np.array(historial)

w_gd, b_gd, hist = descenso_gradiente(x, y, alpha=0.01, iteraciones=10_000)
print(f'w = {w_gd:.3f}  b = {b_gd:.3f}')
print(f'J final = {hist[-1, 2]:.6f}')
```

Mismo resultado que la solución cerrada. Veamos **cómo** llegó: la curva de aprendizaje (costo vs. iteraciones) y la trayectoria recorrida sobre las curvas de nivel.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))

ax[0].plot(hist[:, 2], color=INDIGO)
ax[0].set_xscale('log')
ax[0].axhline(costo(0.6, 2.2, x, y), color=CORAL, ls='--', label='J mínimo')
ax[0].set_xlabel('iteración (escala log)'); ax[0].set_ylabel('J(w, b)')
ax[0].set_title('Curva de aprendizaje'); ax[0].legend()

ax[1].contourf(W, B, J, levels=30, cmap='Purples')
ax[1].plot(hist[::50, 0], hist[::50, 1], '.-', color=INDIGO, ms=3, lw=1, label='trayectoria')
ax[1].scatter(0, 0, color=TINTA, s=40, label='inicio (0, 0)', zorder=3)
ax[1].scatter(0.6, 2.2, color=CORAL, s=90, marker='*', label='mínimo', zorder=3)
ax[1].set_xlabel('w'); ax[1].set_ylabel('b'); ax[1].set_title('Trayectoria en el espacio de parámetros')
ax[1].legend()
plt.tight_layout(); plt.show()
```

Dos observaciones importantes:

- **La curva de aprendizaje baja rápido al principio y luego se aplana.** Casi toda la mejora ocurre en las primeras cientos de iteraciones; el resto es ajuste fino.
- **La trayectoria no va en línea recta al mínimo.** En las primeras iteraciones avanza casi solo en $w$ (la dirección más empinada del cuenco) hasta pasar de largo a $w \approx 1.1$; después gira y recorre lentamente un "valle" alargado hasta el mínimo. Ese valle aparece porque $w$ y $b$ están en escalas distintas (los $x$ valen hasta 5, el término de $b$ vale siempre 1). Lo resolveremos en la sección 7 con el **escalado**.

```python
# ¿Cuántas iteraciones hacían falta realmente?
J_min = costo(w_cerrada, b_cerrada, x, y)
for tol in [1e-2, 1e-4, 1e-6]:
    it = np.argmax(hist[:, 2] - J_min < tol)
    print(f'J − J_min < {tol:g}  alcanzado en la iteración {it}')
```

## 6. El efecto de la tasa de aprendizaje α

$\alpha$ es un **hiperparámetro**: no se aprende de los datos, lo elegimos nosotros. Controla el tamaño de cada paso:

- $\alpha$ **muy pequeño** → pasos diminutos, convergencia lentísima.
- $\alpha$ **adecuado** → convergencia rápida y estable.
- $\alpha$ **demasiado grande** → cada paso "salta" al otro lado del cuenco, más lejos que antes: el algoritmo **diverge**.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
for a, c in [(0.001, '#A5B4FC'), (0.01, INDIGO), (0.1, PURPURA), (0.18, CORAL)]:
    _, _, h = descenso_gradiente(x, y, alpha=a, iteraciones=300)
    ax[0 if a < 0.18 else 1].plot(h[:, 2], color=c, lw=2, label=f'α = {a}')

ax[0].set_title('α que convergen (300 iteraciones)'); ax[0].set_yscale('log')
ax[0].set_xlabel('iteración'); ax[0].set_ylabel('J'); ax[0].legend()
ax[1].set_title('α demasiado grande: divergencia'); ax[1].set_yscale('log')
ax[1].set_xlabel('iteración'); ax[1].set_ylabel('J'); ax[1].legend()
plt.tight_layout(); plt.show()
```

<div style="border-left:4px solid #6B7280; background:#F9FAFB; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#6B7280; font-weight:bold;">NOTA</span> <b>¿Dónde está el límite?</b><br>Para este problema cuadrático el límite puede calcularse: el descenso converge si α &lt; 2/λ<sub>máx</sub>, donde λ<sub>máx</sub> es el mayor autovalor de la matriz hessiana de <i>J</i>. Aquí λ<sub>máx</sub> ≈ 11.83, de modo que el límite es α ≈ 0.169. Con 0.18 ya estamos por encima: la próxima celda lo comprueba. En la práctica no se calcula; se prueban valores en escala logarítmica (0.001, 0.003, 0.01, 0.03, …) y se observa la curva de aprendizaje.</div>

```python
H = np.array([[np.mean(x**2), np.mean(x)],     # Hessiana de J respecto a (w, b)
              [np.mean(x),    1.0       ]])
lam = np.linalg.eigvalsh(H)
print('Autovalores de la hessiana:', lam)
print(f'α máximo para converger    : 2/λmáx = {2/lam.max():.4f}')
print(f'Número de condición λmáx/λmín = {lam.max()/lam.min():.1f}')
```

El **número de condición** (≈ 70) mide qué tan alargado es el cuenco. Cuanto mayor es, más estrecho el valle y más lenta la convergencia: α debe ser pequeño para no divergir en la dirección empinada, pero entonces avanza despacio en la dirección plana. Esto conecta directamente con la siguiente sección.

## 7. Escalado de características

Si **estandarizamos** $x$ (le restamos la media y dividimos por la desviación estándar), la hessiana se vuelve prácticamente diagonal y con autovalores similares: el cuenco pasa de ser un valle alargado a ser casi **circular**.

$$x_{\text{esc}} = \frac{x - \bar{x}}{\sigma_x}$$

```python
mu, sigma = x.mean(), x.std()
x_esc = (x - mu) / sigma
print('x escalado:', x_esc)

H_esc = np.array([[np.mean(x_esc**2), np.mean(x_esc)], [np.mean(x_esc), 1.0]])
print('Autovalores tras escalar:', np.linalg.eigvalsh(H_esc), ' -> número de condición = 1')

# Con el cuenco circular podemos usar un α mucho mayor
w_e, b_e, hist_esc = descenso_gradiente(x_esc, y, alpha=0.5, iteraciones=50)
it_esc = np.argmax(hist_esc[:, 2] - J_min < 1e-6)
it_sin = np.argmax(hist[:, 2] - J_min < 1e-6)
print(f'\nIteraciones hasta J − J_min < 1e-6:  sin escalar = {it_sin},  escalado = {it_esc}')
```

Pasamos de miles de iteraciones a apenas unas pocas. Pero atención: los parámetros aprendidos corresponden a la **variable escalada**, no a la original. Hay que "des-escalarlos" para interpretarlos:

$$\hat{y} = w_{e}\,\frac{x-\bar{x}}{\sigma} + b_{e} \;\Longrightarrow\; w = \frac{w_{e}}{\sigma}, \qquad b = b_{e} - \frac{w_{e}\,\bar{x}}{\sigma}$$

```python
print(f'Parámetros en escala estandarizada: w_e = {w_e:.4f}, b_e = {b_e:.4f}')
w_orig = w_e / sigma
b_orig = b_e - w_e * mu / sigma
print(f'Parámetros en escala original     : w   = {w_orig:.3f}, b   = {b_orig:.3f}')
```

<div style="border-left:4px solid #F43F5E; background:#FFF1F2; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#F43F5E; font-weight:bold;">ADVERTENCIA</span> <b>Escalar con estadísticas de entrenamiento</b><br>La media y la desviación estándar se calculan <b>solo con los datos de entrenamiento</b> y se reutilizan para transformar cualquier dato nuevo. Calcularlas sobre todo el conjunto (incluido el de prueba) filtra información y da métricas optimistas. En scikit-learn esto lo gestiona <code>StandardScaler</code> dentro de un <code>Pipeline</code>.</div>

## 8. Nuestra propia clase, con la API de scikit-learn

Reunimos todo lo aprendido en una clase que **imita la interfaz de scikit-learn** (`fit`, `predict`, `coef_`, `intercept_`). Además, la escribimos **vectorizada**: en lugar de un único $w$, trabaja con un vector de pesos $\mathbf{w}$, por lo que sirve para cualquier número de características.

Con $m$ muestras y $n$ características, las fórmulas se vuelven matriciales:

$$\hat{\mathbf{y}} = X\mathbf{w} + b, \qquad \nabla_{\mathbf{w}} J = \frac{1}{m}X^{\top}(\hat{\mathbf{y}} - \mathbf{y}), \qquad \frac{\partial J}{\partial b} = \frac{1}{m}\sum_i (\hat{y}^{(i)} - y^{(i)})$$

```python
class RegresionLinealGD:
    """Regresión lineal por descenso de gradiente, con escalado interno opcional."""

    def __init__(self, alpha=0.1, iteraciones=1000, escalar=True, tol=1e-10):
        self.alpha, self.iteraciones = alpha, iteraciones
        self.escalar, self.tol = escalar, tol

    def fit(self, X, y):
        X = np.asarray(X, dtype=float); y = np.asarray(y, dtype=float)
        m, n = X.shape
        # Escalado (estadísticas guardadas para usarlas después)
        self.mu_ = X.mean(axis=0) if self.escalar else np.zeros(n)
        self.sigma_ = X.std(axis=0) if self.escalar else np.ones(n)
        Xs = (X - self.mu_) / self.sigma_

        w, b = np.zeros(n), 0.0
        self.historial_ = []
        for it in range(self.iteraciones):
            error = Xs @ w + b - y
            dw = Xs.T @ error / m
            db = error.mean()
            w -= self.alpha * dw
            b -= self.alpha * db
            J = (error ** 2).mean() / 2
            self.historial_.append(J)
            if it > 0 and abs(self.historial_[-2] - J) < self.tol:   # criterio de parada
                break
        self.n_iter_ = it + 1

        # Convertimos los parámetros a la escala original de X
        self.coef_ = w / self.sigma_
        self.intercept_ = b - np.sum(w * self.mu_ / self.sigma_)
        return self

    def predict(self, X):
        return np.asarray(X, dtype=float) @ self.coef_ + self.intercept_

    def score(self, X, y):
        return r2_score(y, self.predict(X))


mi_modelo = RegresionLinealGD(alpha=0.5).fit(X, y)
print(f'Nuestra clase : w = {mi_modelo.coef_[0]:.3f}  b = {mi_modelo.intercept_:.3f}  ({mi_modelo.n_iter_} iteraciones)')
print(f'scikit-learn  : w = {modelo.coef_[0]:.3f}  b = {modelo.intercept_:.3f}')
```

<div style="border-left:4px solid #4F46E5; background:#EEF2FF; padding:10px 14px; margin:8px 0; border-radius:4px; color:#111827;"><span style="font-family:monospace; font-size:0.8em; color:#4F46E5; font-weight:bold;">IDEA CLAVE</span> <b>Diseño con la misma API</b><br>Al respetar la interfaz <code>fit / predict / coef_ / intercept_</code>, nuestra clase es intercambiable con la de scikit-learn en el resto del código. Además incorporamos un <b>criterio de parada</b>: si el costo deja de mejorar (cambio menor que <code>tol</code>), no tiene sentido seguir iterando.</div>

## 9. Un caso más realista: varias características

Generamos un conjunto sintético de 200 estudiantes con **tres** características y una relación lineal conocida más ruido. Como conocemos los coeficientes "verdaderos", podemos comprobar si los métodos los recuperan.

| Característica | Rango | Coeficiente real |
|---|---|---|
| horas de estudio por semana | 0 – 20 | +2.5 |
| asistencia (%) | 50 – 100 | +0.3 |
| horas de redes sociales por día | 0 – 8 | −1.8 |
| intercepto | — | 20 |

```python
rng = np.random.default_rng(672)
m = 200
horas   = rng.uniform(0, 20, m)
asist   = rng.uniform(50, 100, m)
redes   = rng.uniform(0, 8, m)
X_multi = np.column_stack([horas, asist, redes])

w_real, b_real = np.array([2.5, 0.3, -1.8]), 20.0
y_multi = X_multi @ w_real + b_real + rng.normal(0, 4, m)   # ruido gaussiano σ = 4

# División entrenamiento / prueba (80 / 20)
from sklearn.model_selection import train_test_split
X_tr, X_te, y_tr, y_te = train_test_split(X_multi, y_multi, test_size=0.2, random_state=0)
print('Entrenamiento:', X_tr.shape, '  Prueba:', X_te.shape)
```

```python
sk = LinearRegression().fit(X_tr, y_tr)
gd = RegresionLinealGD(alpha=0.1, iteraciones=5000).fit(X_tr, y_tr)
gd_sin = RegresionLinealGD(alpha=0.0002, iteraciones=5000, escalar=False).fit(X_tr, y_tr)

nombres = ['horas', 'asistencia', 'redes', 'intercepto']
print(f'{"":12}{"real":>9}{"sklearn":>10}{"GD escal.":>11}{"GD sin esc.":>13}')
for i, nom in enumerate(nombres):
    r = w_real[i] if i < 3 else b_real
    s = sk.coef_[i] if i < 3 else sk.intercept_
    g = gd.coef_[i] if i < 3 else gd.intercept_
    g2 = gd_sin.coef_[i] if i < 3 else gd_sin.intercept_
    print(f'{nom:12}{r:9.3f}{s:10.3f}{g:11.3f}{g2:13.3f}')

print(f'\nIteraciones — GD escalado: {gd.n_iter_},  GD sin escalar: {gd_sin.n_iter_}')
print(f'R² en prueba — sklearn: {sk.score(X_te, y_te):.4f}   GD escalado: {gd.score(X_te, y_te):.4f}   GD sin escalar: {gd_sin.score(X_te, y_te):.4f}')
```

Tres lecturas de esta tabla:

1. **scikit-learn y nuestro GD escalado coinciden** hasta el tercer decimal, y ambos se acercan a los pesos reales. No los reproducen exactamente porque los datos tienen ruido: con 160 muestras hay incertidumbre en la estimación. El intercepto es el que más se desvía (≈ 24 frente a 20) porque representa la nota con asistencia 0 %, un valor muy lejos de los datos (50–100 %): es una **extrapolación**, y pequeños errores en la pendiente de la asistencia se amplifican al llevar la recta hasta allí.
2. **Sin escalar, el descenso de gradiente fracasa en la práctica.** La asistencia (valores ~75) domina la escala; para no divergir hay que usar un α diminuto, y tras 5000 iteraciones el intercepto sigue casi en 0 y el peso de la asistencia está inflado para compensarlo. Lo engañoso es que su R² (≈ 0.90) parece aceptable: **un buen R² no garantiza que los coeficientes sean correctos**, y si los usáramos para interpretar el efecto de cada variable llegaríamos a conclusiones falsas.
3. Evaluamos en datos de **prueba** que el modelo no vio durante el entrenamiento: esa es la medida honesta de su capacidad de generalización.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4.2))
ax[0].plot(gd.historial_, color=INDIGO, lw=2, label=f'escalado (α=0.1)')
ax[0].plot(gd_sin.historial_, color=CORAL, lw=2, label=f'sin escalar (α=0.0002)')
ax[0].set_yscale('log'); ax[0].set_xscale('log')
ax[0].set_xlabel('iteración'); ax[0].set_ylabel('J'); ax[0].set_title('Curvas de aprendizaje'); ax[0].legend()

y_hat = sk.predict(X_te)
lim = [y_te.min() - 5, y_te.max() + 5]
ax[1].scatter(y_te, y_hat, color=PURPURA, alpha=0.7)
ax[1].plot(lim, lim, color=TINTA, ls='--', lw=1, label='predicción perfecta')
ax[1].set_xlabel('y real'); ax[1].set_ylabel('y predicha'); ax[1].set_title('Real vs. predicho (conjunto de prueba)')
ax[1].legend()
plt.tight_layout(); plt.show()
```

En el gráfico de la derecha, cuanto más cerca están los puntos de la diagonal, mejor la predicción. La dispersión alrededor de ella corresponde principalmente al ruido que introdujimos ($\sigma = 4$), que ningún modelo puede eliminar.

Finalmente, la **interpretación** de los coeficientes es directa gracias a que el modelo es lineal:

```python
for nom, c in zip(nombres[:3], sk.coef_):
    signo = 'aumenta' if c > 0 else 'disminuye'
    print(f'Por cada unidad adicional de {nom:<10} la nota {signo} {abs(c):.2f} puntos (manteniendo lo demás constante).')
```

## 10. Conclusiones

| Aspecto | Solución cerrada / scikit-learn | Descenso de gradiente |
|---|---|---|
| Idea | Despejar el mínimo analíticamente | Bajar por la superficie de $J$ paso a paso |
| Resultado | Exacto en una sola operación | Aproximado, mejora con cada iteración |
| Hiperparámetros | Ninguno | $\alpha$, número de iteraciones, tolerancia |
| Sensible al escalado | No | **Sí**, mucho |
| Costo con $n$ grande | Alto ($\sim n^3$) | Bajo por iteración ($\sim m\,n$) |
| ¿Sirve para otros modelos? | Solo casos particulares | **Sí**: logística, redes neuronales, etc. |

Los dos caminos llegan al mismo par $(w, b) = (0.6,\ 2.2)$ en el ejemplo simple, y a los mismos coeficientes en el caso multivariable. La ecuación normal es la herramienta adecuada para la regresión lineal con pocas características; el **descenso de gradiente** es la idea que reaparecerá en todo el resto del curso, porque la mayoría de los modelos no tiene solución cerrada.
