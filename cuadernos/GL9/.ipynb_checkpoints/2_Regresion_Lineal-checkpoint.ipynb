En este cuaderno se explican los fundamentos de la **regresión lineal** aplicada a un
caso sencillo de **Ciberseguridad**: estimar cuánto costaría un incidente a partir del número
de equipos afectados. Recorreremos todo el flujo de trabajo típico de un proyecto de
Machine Learning supervisado: generar los datos, explorarlos, preparar el modelo, entrenarlo,
evaluarlo y usarlo para predecir.

<div style="border-left:4px solid #7E22CE;background:rgba(126,34,206,0.07);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#7E22CE;font-weight:600;margin-bottom:4px;">¿Qué es la regresión lineal?</div><div style="font-size:15px;line-height:1.6;">La <b>regresión lineal simple</b> busca una recta que describa la relación entre una
variable de entrada <i>x</i> (aquí, los equipos afectados) y una variable de salida continua
<i>y</i> (aquí, el coste). La recta se expresa como la <b>función hipótesis</b>:
<div style="text-align:center;font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:16px;margin:10px 0;color:#7E22CE;">
h(x) = &theta;<sub>0</sub> + &theta;<sub>1</sub> &middot; x</div>
donde <b>&theta;<sub>0</sub></b> es la ordenada al origen (el coste base cuando x = 0) y
<b>&theta;<sub>1</sub></b> es la pendiente (cuánto sube el coste por cada equipo adicional).
Entrenar el modelo consiste en <b>encontrar los valores de &theta; que mejor ajustan los datos</b>,
minimizando el error cuadrático entre las predicciones y los valores reales.</div></div>

## Enunciado del ejercicio

El objetivo es **predecir el coste de un incidente de seguridad** en base al **número de
equipos que se han visto afectados**. Como no disponemos de datos reales, generaremos un
conjunto de datos sintético de forma aleatoria pero con una tendencia lineal conocida, lo que
nos permitirá comprobar después si el modelo "redescubre" esa tendencia.

<div style="border-left:4px solid #6B7280;background:rgba(107,114,128,0.08);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#6B7280;font-weight:600;margin-bottom:4px;">Flujo del cuaderno</div><div style="font-size:15px;line-height:1.6;"><b>1.</b> Generación de datos &rarr; <b>2.</b> Visualización &rarr; <b>3.</b> Preparación
(escalado) &rarr; <b>4.</b> Construcción del modelo &rarr; <b>5.</b> Evaluación &rarr;
<b>6.</b> Predicción de nuevos ejemplos.</div></div>

### 1. Generación del conjunto de datos

Creamos 100 ejemplos sintéticos. La entrada `X` se toma aleatoriamente en el rango [0, 2]
y el coste `y` se construye a partir de una **relación lineal verdadera** más un poco de
**ruido aleatorio** que simula la variabilidad del mundo real:
<div style="text-align:center;font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:15px;margin:10px 0;color:#111827;">
y = 4 + 3&middot;X + ruido</div>
Es decir, los parámetros "reales" que ocultamos al modelo son &theta;<sub>0</sub> = 4 y
&theta;<sub>1</sub> = 3. Más adelante veremos si el modelo se acerca a ellos.

```python
import numpy as np

# Semilla fija: garantiza que los datos (y por tanto los resultados) sean reproducibles
# en cada ejecución del cuaderno.
np.random.seed(42)

# X: numero de equipos afectados (a escala reducida, rango [0, 2])
X = 2 * np.random.rand(100, 1)

# y: coste del incidente = relacion lineal verdadera (4 + 3*X) + ruido gaussiano
y = 4 + 3 * X + np.random.randn(100, 1)

print("La longitud del conjunto de datos es:", len(X))
```

<div style="border-left:4px solid #4F46E5;background:rgba(79,70,229,0.07);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#4F46E5;font-weight:600;margin-bottom:4px;">¿Por qué fijar la semilla?</div><div style="font-size:15px;line-height:1.6;">`np.random.rand` y `np.random.randn` producen números <b>distintos en cada ejecución</b>.
Sin `np.random.seed(42)`, los valores de &theta; y las gráficas cambiarían cada vez que
corras el cuaderno, y las explicaciones dejarían de coincidir con las salidas. Fijar la
semilla hace el experimento <b>reproducible</b>: obtendrás siempre los mismos resultados.</div></div>

### 2. Visualización del conjunto de datos

Antes de modelar, siempre conviene **mirar los datos**. Una nube de puntos nos dice de un
vistazo si la relación entre entrada y salida es aproximadamente lineal, lo que justifica usar
regresión lineal.

```python
import matplotlib.pyplot as plt
%matplotlib inline

# Colores de la identidad visual de la materia (INF672)
C_DATOS   = "#4F46E5"   # indigo  -> puntos de datos
C_MODELO  = "#7E22CE"   # purpura -> recta del modelo
C_NUEVO   = "#F43F5E"   # coral   -> nuevas predicciones
```

Una primera aproximación, sin etiquetas todavía:

```python
plt.plot(X, y, ".", color=C_DATOS)
plt.title("Datos crudos (sin escalar)")
plt.show()
```

<div style="border-left:4px solid #6B7280;background:rgba(107,114,128,0.08);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#6B7280;font-weight:600;margin-bottom:4px;">Buena práctica</div><div style="font-size:15px;line-height:1.6;">Una gráfica sin ejes rotulados es difícil de interpretar. Añade siempre <b>etiquetas y
unidades</b> a los ejes.</div></div>

```python
plt.plot(X, y, ".", color=C_DATOS)
plt.xlabel("Equipos afectados (u/1000)")
plt.ylabel("Coste del incidente (u/10000)")
plt.title("Datos crudos con ejes rotulados")
plt.show()
```

### 3. Modificación del conjunto de datos

Los datos generados están en unidades reducidas: `X` en miles de equipos y `y` en decenas
de miles de euros (de ahí las etiquetas "u/1000" y "u/10000"). Para que las cifras sean
**interpretables en unidades reales**, las llevamos a su escala: multiplicamos los equipos por
1000 y el coste por 10000, y los convertimos a enteros.

```python
import pandas as pd
```

```python
# Construimos un DataFrame de pandas a partir de los arrays de NumPy.
# .flatten() convierte las columnas (100,1) en vectores (100,) de una dimension.
data = {'n_equipos_afectados': X.flatten(), 'coste': y.flatten()}
df = pd.DataFrame(data)
df.head(10)
```

```python
# Escalado del numero de equipos afectados (a unidades reales) y conversion a entero
df['n_equipos_afectados'] = (df['n_equipos_afectados'] * 1000).astype('int')
# Escalado del coste (a euros) y conversion a entero
df['coste'] = (df['coste'] * 10000).astype('int')
df.head(10)
```

```python
# Representacion grafica del conjunto de datos ya escalado
plt.plot(df['n_equipos_afectados'], df['coste'], ".", color=C_DATOS)
plt.xlabel("Equipos afectados")
plt.ylabel("Coste del incidente (€)")
plt.title("Datos escalados a unidades reales")
plt.show()
```

<div style="border-left:4px solid #F43F5E;background:rgba(244,63,94,0.07);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#F43F5E;font-weight:600;margin-bottom:4px;">El escalado no cambia la relación</div><div style="font-size:15px;line-height:1.6;">La forma de la nube de puntos es <b>idéntica</b> a la anterior: solo han cambiado los números
de los ejes. El escalado afecta el <i>valor</i> de los parámetros &theta; (la pendiente y la
ordenada), pero no la calidad del ajuste ni la relación lineal subyacente.</div></div>

### 4. Construcción del modelo

Usamos `LinearRegression` de scikit-learn. El método `.fit()` recibe las entradas y las
salidas de entrenamiento y calcula, mediante **mínimos cuadrados ordinarios (OLS)**, los
valores de &theta;<sub>0</sub> y &theta;<sub>1</sub> que minimizan el error cuadrático medio.

```python
from sklearn.linear_model import LinearRegression
```

```python
# Construccion del modelo y ajuste (entrenamiento) de la funcion hipotesis
lin_reg = LinearRegression()

# scikit-learn espera la entrada X como matriz 2D (n_muestras, n_caracteristicas).
# reshape(-1, 1) convierte el vector de equipos en una matriz columna.
lin_reg.fit(df['n_equipos_afectados'].values.reshape(-1, 1), df['coste'].values)
```

Una vez entrenado, el modelo expone los parámetros aprendidos:

```python
# Parametro theta_0 (ordenada al origen): coste estimado cuando hay 0 equipos afectados
lin_reg.intercept_
```

```python
# Parametro theta_1 (pendiente): incremento de coste por cada equipo afectado adicional
lin_reg.coef_
```

<div style="border-left:4px solid #16A34A;background:rgba(22,163,74,0.07);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#16A34A;font-weight:600;margin-bottom:4px;">¿Reconoció el modelo la tendencia real?</div><div style="font-size:15px;line-height:1.6;">Recuerda que generamos los datos con &theta;<sub>0</sub> = 4 y &theta;<sub>1</sub> = 3 en la
escala reducida. Tras multiplicar los equipos por 1000 y el coste por 10000, los parámetros
"verdaderos" en unidades reales son:
<div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace; margin:8px 0;">
&theta;<sub>0</sub> &asymp; 4 &times; 10000 = <b>40&nbsp;000</b> &nbsp;&nbsp;|&nbsp;&nbsp;
&theta;<sub>1</sub> &asymp; 3 &times; (10000 / 1000) = <b>30</b></div>
El modelo aprende valores cercanos (aprox. &theta;<sub>0</sub> &asymp; 42&nbsp;000 y
&theta;<sub>1</sub> &asymp; 28). No son exactos porque el <b>ruido</b> y la <b>conversión a
enteros</b> distorsionan ligeramente los datos: eso es justo lo que ocurre con datos reales.</div></div>

Con &theta;<sub>0</sub> y &theta;<sub>1</sub> ya podemos dibujar la **recta de regresión**.
Como una recta queda definida por dos puntos, basta con predecir el coste en el valor mínimo y
en el valor máximo de equipos y unirlos.

```python
# Prediccion para el valor minimo y maximo del conjunto de entrenamiento
X_min_max = np.array([[df["n_equipos_afectados"].min()],
                      [df["n_equipos_afectados"].max()]])
y_train_pred = lin_reg.predict(X_min_max)
```

```python
# Representacion grafica de la funcion hipotesis (recta) sobre los datos
plt.plot(X_min_max, y_train_pred, "-", color=C_MODELO, linewidth=2, label="Modelo h(x)")
plt.plot(df['n_equipos_afectados'], df['coste'], ".", color=C_DATOS, label="Datos")
plt.xlabel("Equipos afectados")
plt.ylabel("Coste del incidente (€)")
plt.title("Recta de regresión ajustada")
plt.legend()
plt.show()
```

### 5. Evaluación del modelo

Ajustar una recta no basta: hay que medir **qué tan bien** describe los datos. Una métrica
habitual en regresión es el **coeficiente de determinación R²**, que indica la proporción de la
variabilidad del coste que el modelo logra explicar. Va de 0 a 1: cuanto más cerca de 1, mejor.

```python
from sklearn.metrics import r2_score

y_pred_train = lin_reg.predict(df['n_equipos_afectados'].values.reshape(-1, 1))
r2 = r2_score(df['coste'].values, y_pred_train)
print(f"R² del modelo sobre los datos de entrenamiento: {r2:.4f}")
```

<div style="border-left:4px solid #6B7280;background:rgba(107,114,128,0.08);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#6B7280;font-weight:600;margin-bottom:4px;">Interpretación del R²</div><div style="font-size:15px;line-height:1.6;">Un R² de aproximadamente <b>0.77</b> significa que el modelo explica cerca del <b>77&nbsp;%</b>
de la variación del coste. El 23&nbsp;% restante corresponde al ruido que añadimos a propósito
al generar los datos, que ningún modelo lineal puede capturar.</div></div>

### 6. Predicción de nuevos ejemplos

El objetivo final del modelo es **predecir el coste de incidentes futuros**. Estimemos, por
ejemplo, cuánto costaría un incidente que afecte a 1300 equipos.

```python
x_new = np.array([[1300]])   # 1300 equipos afectados

# Prediccion del coste del incidente
coste = lin_reg.predict(x_new)

coste_fmt = f"{int(round(coste[0])):,}".replace(",", ".")
print(f"El coste estimado del incidente sería: {coste_fmt} €")
```

```python
# Situamos la nueva prediccion (coral) sobre los datos y la recta del modelo
plt.plot(df['n_equipos_afectados'], df['coste'], ".", color=C_DATOS, label="Datos")
plt.plot(X_min_max, y_train_pred, "-", color=C_MODELO, linewidth=2, label="Modelo h(x)")
plt.plot(x_new, coste, "X", color=C_NUEVO, markersize=12, label="Predicción (1300)")
plt.xlabel("Equipos afectados")
plt.ylabel("Coste del incidente (€)")
plt.title("Predicción de un nuevo ejemplo")
plt.legend()
plt.show()
```

<div style="border-left:4px solid #4F46E5;background:rgba(79,70,229,0.07);padding:12px 16px;margin:6px 0;border-radius:0 8px 8px 0;font-family:'IBM Plex Sans','Segoe UI',system-ui,-apple-system,sans-serif;color:#111827;"><div style="font-family:'IBM Plex Mono','Consolas','Courier New',monospace;font-size:12px;letter-spacing:.08em;text-transform:uppercase;color:#4F46E5;font-weight:600;margin-bottom:4px;">Interpolación vs. extrapolación</div><div style="font-size:15px;line-height:1.6;">El valor 1300 está <b>dentro</b> del rango de entrenamiento (aprox. 11 a 1973 equipos), por lo
que se trata de una <b>interpolación</b> fiable. Predecir muy por fuera de ese rango
(p. ej. 10&nbsp;000 equipos) sería <b>extrapolación</b>: el modelo devolvería un número, pero no
hay datos que respalden que la relación siga siendo lineal ahí. Interpreta esas predicciones
con cautela.</div></div>

## Conclusiones

- La regresión lineal ajusta una recta `h(x) = θ₀ + θ₁·x` a los datos minimizando el error cuadrático.
- El modelo recuperó parámetros cercanos a la tendencia real que ocultamos al generar los datos.
- El R² cuantifica la bondad del ajuste; el ruido pone un techo a lo que un modelo lineal puede lograr.
- Las predicciones son confiables dentro del rango de entrenamiento (interpolación) y arriesgadas fuera de él (extrapolación).
