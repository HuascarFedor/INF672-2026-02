# De los datos a los gráficos: NumPy, Pandas, Matplotlib y Seaborn

Este cuaderno recorre, paso a paso, las cuatro librerías que forman el núcleo del análisis de datos en Python. La idea es ir **de lo más simple a lo más aplicado**:

1. **NumPy:** cómo se guardan y operan los números (el arreglo `ndarray`).
2. **scikit-learn:** de dónde sacamos un conjunto de datos real (*Iris*).
3. **Pandas:** cómo organizamos esos datos en tablas con nombres (`Series` y `DataFrame`).
4. **Matplotlib y Seaborn:** cómo los convertimos en gráficos que se entienden.

Usaremos siempre el mismo dataset —**Iris**, 150 flores medidas en 4 dimensiones— para que veas cómo un mismo conjunto de datos se recorre, se filtra, se resume y se grafica.

---

## Librerías y su rol

Antes de empezar cargamos las herramientas. Cada `import` trae una pieza distinta:

| Librería | Alias | Para qué sirve |
|----------|-------|----------------|
| `numpy`  | `np`  | Cálculo numérico con arreglos y matrices. |
| `pandas` | `pd`  | Tablas de datos con etiquetas (filas y columnas). |
| `matplotlib.pyplot` | `plt` | Gráficos base (dispersión, barras, histogramas). |
| `seaborn` | `sns` | Gráficos estadísticos con estilo, sobre matplotlib. |

Los **alias** (`np`, `pd`, `plt`, `sns`) son una convención universal: hacen el código más corto y todo el mundo los reconoce.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. NumPy: el arreglo (`ndarray`)

El `ndarray` es la estructura central de NumPy: una colección de números **del mismo tipo**, guardados de forma contigua en memoria. Esto lo hace mucho más rápido que una lista de Python para operaciones matemáticas.

### 1.1 Crear un arreglo a mano

`np.array([...])` convierte una lista de Python en un arreglo de NumPy.

```python
a = np.array([2, 4, 6, 8])
a
```

### 1.2 Crear arreglos con un patrón

A menudo no queremos escribir los valores uno por uno, sino generarlos con una regla.

**`np.arange(inicio, fin, paso)`** funciona como el `range` de Python pero devuelve un arreglo: genera valores desde `inicio` hasta *justo antes* de `fin`, saltando de `paso` en `paso`.

```python
np.arange(0, 10, 2)      # inicio, fin (excl.), paso
```

**`np.linspace(inicio, fin, n)`** reparte `n` valores **equiespaciados** entre `inicio` y `fin`, incluyendo ambos extremos. Muy útil para crear ejes o muestrear una función de forma uniforme.

👉 Diferencia clave: en `arange` tú controlas el **paso**; en `linspace` controlas la **cantidad** de puntos.

```python
np.linspace(0, 1, 5)     # 5 valores equiespaciados en [0, 1]
```

## 2. Matrices e indexación

Un arreglo puede tener más de una dimensión. Una lista de listas se convierte en una **matriz** (arreglo 2D): filas y columnas.

Tres atributos describen cualquier arreglo:

- **`.shape`** → forma, `(filas, columnas)`.
- **`.dtype`** → tipo de dato de los elementos (`int64`, `float64`, …).
- **`.ndim`** → número de dimensiones (1 = vector, 2 = matriz).

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])
M.shape, M.dtype, M.ndim
```

### 2.1 Seleccionar filas

La indexación empieza en **0**. `M[0]` devuelve la primera fila completa como un arreglo 1D.

```python
M[0]        # primera fila
```

### 2.2 Seleccionar columnas

La coma separa las dimensiones: `M[filas, columnas]`. Los dos puntos `:` significan «todas». Así, `M[:, 1]` se lee *«todas las filas, columna 1»* → la segunda columna.

```python
M[:, 1]     # segunda columna (todas las filas)
```

## 3. Operaciones vectorizadas y agregaciones

La gran ventaja de NumPy: las operaciones se aplican a **todos los elementos a la vez**, sin escribir bucles. A esto se le llama **vectorización**, y es lo que hace al código corto y rápido.

`a * 10` multiplica cada elemento por 10 de una sola vez.

```python
a * 10          # multiplica cada elemento
```

### 3.1 Estadísticos resumen

Los arreglos traen métodos para resumir sus valores en un solo número:

- **`.mean()`** → promedio.
- **`.sum()`** → suma total.
- **`.std()`** → desviación estándar (cuánto se dispersan los datos respecto a la media).

```python
a.mean(), a.sum(), a.std()
```

### 3.2 El parámetro `axis`

En una matriz podemos resumir **por filas** o **por columnas**. Eso lo decide `axis`:

- **`axis=0`** → recorre las filas → resultado **por columna**.
- **`axis=1`** → recorre las columnas → resultado **por fila**.

`M.sum(axis=0)` suma cada columna hacia abajo.

```python
M.sum(axis=0)   # suma por columnas (recorre las filas)
```

## 4. Un dataset real: *Iris* con scikit-learn

Hasta aquí usamos números inventados. Ahora traemos un conjunto de datos clásico: **Iris**, con medidas de 150 flores de tres especies.

`load_iris()` devuelve un objeto con los datos. `iris.data` es una matriz NumPy de **150 filas (muestras) × 4 columnas (características)**: largo y ancho de sépalo, largo y ancho de pétalo.

```python
from sklearn.datasets import load_iris
 
iris = load_iris()
iris.data.shape          # 150 muestras, 4 características
```

Como `iris.data` es un arreglo NumPy, podemos aplicarle directamente lo que ya aprendimos. `iris.data.mean(axis=0)` calcula la **media de cada una de las 4 características** (promedio por columna).

```python
iris.data.mean(axis=0)   # media de cada característica
```

## 5. Pandas: `Series`

Una **`Series`** es un arreglo de una dimensión pero con **etiquetas** (un índice). Es como un arreglo de NumPy al que le ponemos nombre a cada posición.

Aquí las medidas se etiquetan como `m1`, `m2`, … en lugar de usar solo posiciones numéricas.

```python
s = pd.Series([5.1, 4.9, 4.7, 4.6],
              index=["m1", "m2", "m3", "m4"])
s
```

### 5.1 Acceso por etiqueta y por posición

Con etiquetas tenemos dos formas de acceder a un dato:

- **`s["m1"]`** → por **etiqueta** (el nombre del índice).
- **`s.iloc[0]`** → por **posición** entera (`iloc` = *integer location*).

Ambas devuelven el mismo valor, pero conviene distinguirlas: la etiqueta no siempre coincide con la posición.

```python
s["m1"], s.iloc[0]      # por etiqueta y por posición
```

### 5.2 Filtrado booleano

`s > 4.7` genera una máscara de `True`/`False` para cada elemento. Al ponerla entre corchetes, `s[s > 4.7]` **conserva solo los elementos que cumplen la condición**. Es una de las herramientas más usadas en el análisis de datos.

```python
s[s > 4.7]              # filtrado booleano
```

### 5.3 Agregaciones sobre la Series

Igual que los arreglos, las `Series` traen métodos de resumen como `.mean()` y `.max()`.

```python
s.mean(), s.max()
```

## 6. Pandas: `DataFrame`

El **`DataFrame`** es la estructura estrella de Pandas: una **tabla** con filas y columnas etiquetadas, donde cada columna puede tener su propio tipo de dato. Piensa en una hoja de cálculo dentro de Python.

### 6.1 Crear un DataFrame desde un diccionario

La forma más directa: cada **clave** del diccionario es el nombre de una columna, y cada **lista** son sus valores.

```python
pd.DataFrame({
    "nombre": ["Ana", "Beto", "Carla"],
    "nota":   [89, 74, 95],
})
```

### 6.2 Iris como DataFrame

Ahora convertimos el dataset Iris en una tabla con nombres de columna legibles. Además añadimos la columna **`species`** con la especie de cada flor, usando `Categorical.from_codes` para traducir los códigos numéricos (`0,1,2`) a nombres (`setosa`, `versicolor`, `virginica`).

**`df.head()`** muestra las primeras filas: la forma rápida de «asomarse» a una tabla.

```python
from sklearn.datasets import load_iris
iris = load_iris()
 
df = pd.DataFrame(iris.data,
                  columns=["sepal_len", "sepal_wid",
                           "petal_len", "petal_wid"])
df["species"] = pd.Categorical.from_codes(iris.target,
                                          iris.target_names)
df.head()
```

**`df.shape`** confirma las dimensiones de la tabla: `(filas, columnas)`. Debe darnos 150 flores y 5 columnas (las 4 medidas + la especie).

```python
df.shape
```

### 6.3 Resumen estadístico de una columna

**`.describe()`** entrega de un solo golpe los estadísticos clave de una columna numérica: conteo, media, desviación estándar, mínimo, cuartiles y máximo. Ideal para conocer la distribución de un dato sin graficar todavía.

```python
df["sepal_len"].describe()
```

### 6.4 Filtrar filas por condición

Igual que con las `Series`, `df["species"] == "setosa"` crea una máscara booleana, y `df[...]` devuelve **solo las filas** donde la condición es verdadera. Aquí nos quedamos únicamente con las flores *setosa*.

```python
df[df["species"] == "setosa"].head(3)   # solo setosa
```

### 6.5 Crear columnas derivadas

Podemos calcular una columna nueva a partir de otras. Aquí definimos **`petal_ratio`** como la razón largo/ancho del pétalo. La operación se aplica fila por fila de forma vectorizada, sin bucles.

```python
df["petal_ratio"] = df["petal_len"] / df["petal_wid"]
df[["petal_len", "petal_wid", "petal_ratio"]].head(3)
```

### 6.6 Agrupar y resumir (`groupby`)

**`groupby`** es el patrón *«dividir → aplicar → combinar»*: separa la tabla por los valores de una columna (aquí, la especie), calcula un resumen en cada grupo (la media) y junta los resultados.

El resultado responde: *¿cuál es el largo medio de sépalo y de pétalo en cada especie?* Verás que *setosa* tiene pétalos mucho más pequeños — un dato clave para distinguir especies.

```python
df.groupby("species")[["sepal_len", "petal_len"]].mean()
```

### 6.7 Guardar y leer archivos (CSV)

El análisis casi siempre implica intercambiar datos con archivos:

- **`.to_csv("iris.csv", index=False)`** escribe la tabla a disco (`index=False` evita guardar el índice como columna extra).
- **`pd.read_csv("iris.csv")`** vuelve a cargarla.

El CSV es el formato de intercambio más universal para tablas.

```python
df.to_csv("iris.csv", index=False)   # guardar a archivo
datos = pd.read_csv("iris.csv")      # leer desde archivo
datos.head(2)
```

## 7. Visualización con Matplotlib

Los números resumen; los gráficos **revelan**. Matplotlib es la librería base para graficar en Python.

### 7.1 Gráfico de dispersión (*scatter*)

Un *scatter* coloca un punto por cada fila usando dos variables como coordenadas `(x, y)`. Sirve para ver **relaciones entre dos medidas**.

Aquí enfrentamos largo de sépalo vs. largo de pétalo. Fíjate en la nube de puntos: cuando los pétalos crecen, los sépalos tienden a crecer también → hay **correlación positiva**. Las funciones `xlabel`, `ylabel` y `title` etiquetan el gráfico para que se entienda solo.

```python
plt.scatter(df["sepal_len"], df["petal_len"])
plt.xlabel("Largo de sépalo (cm)")
plt.ylabel("Largo de pétalo (cm)")
plt.title("Iris: sépalo vs pétalo")
plt.show()
```

### 7.2 Histograma

Un histograma agrupa los valores de **una sola variable** en intervalos (`bins`) y cuenta cuántos caen en cada uno. Muestra la **forma de la distribución**: dónde se concentran los datos, si es simétrica, si hay varios picos.

Observa aquí en qué rango de largo de sépalo se acumulan más flores.

```python
plt.hist(df["sepal_len"], bins=15, color="#4F46E5",
         edgecolor="white")
plt.xlabel("Largo de sépalo (cm)")
plt.ylabel("Frecuencia")
plt.show()
```

### 7.3 Gráfico de barras

Primero calculamos con `groupby` el largo medio de pétalo por especie, y luego lo graficamos con `.plot(kind="bar")`. Las barras sirven para **comparar una cantidad entre categorías**.

El contraste es claro: *setosa* tendrá una barra mucho más baja que las otras dos especies.

```python
medias = df.groupby("species")["petal_len"].mean()
medias.plot(kind="bar")
plt.ylabel("Largo de pétalo medio (cm)")
plt.show()
```

### 7.4 Varios gráficos en una figura (*subplots*)

`plt.subplots(1, 2)` crea una figura con **dos ejes lado a lado**. Cada eje (`axes[0]`, `axes[1]`) se dibuja por separado. Es útil para **comparar dos vistas** a la vez: aquí, la relación largo–ancho del sépalo frente a la del pétalo. `figsize` controla el tamaño de la figura en pulgadas.

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].scatter(df["sepal_len"], df["sepal_wid"])
axes[0].set_title("Sépalo")
axes[1].scatter(df["petal_len"], df["petal_wid"])
axes[1].set_title("Pétalo")
plt.show()
```

## 8. Visualización con Seaborn

Seaborn trabaja **sobre** Matplotlib pero entiende directamente los `DataFrame` y produce gráficos estadísticos con mejor estilo y menos código.

### 8.1 Diagrama de caja (*boxplot*)

El *boxplot* resume la distribución de una variable **por categoría**. Cada caja muestra la mediana (línea central), el rango intercuartílico (la caja), los bigotes y los posibles valores atípicos.

Al comparar el largo de pétalo por especie se ve de inmediato **cuál especie tiene pétalos más largos y cuánto varían** dentro de cada grupo.

```python
sns.boxplot(data=df, x="species", y="petal_len")
plt.show()
```

### 8.2 Dispersión coloreada por categoría (`hue`)

Es el mismo scatter de antes, pero con el parámetro **`hue="species"`** que **colorea cada punto según su especie**. Añadir el color convierte un gráfico plano en uno que cuenta una historia: verás que las tres especies forman **grupos separados**, lo que anticipa que son fáciles de clasificar.

```python
sns.scatterplot(data=df, x="sepal_len", y="petal_len",
                hue="species")
plt.show()
```

### 8.3 Mapa de calor de correlaciones (*heatmap*)

Primero `df.corr()` calcula la **matriz de correlación** entre todas las variables numéricas (quitamos `species` porque es texto). Cada valor va de −1 a 1 e indica qué tan relacionadas están dos variables.

`sns.heatmap(..., annot=True)` la dibuja como una cuadrícula de colores con los números encima. Los tonos más intensos marcan las correlaciones más fuertes: aquí destacará la fortísima relación entre **largo y ancho de pétalo**.

---

### ✅ Resumen

Recorrimos el flujo completo del análisis de datos:

**NumPy** (arreglos y operaciones) → **Pandas** (tablas, filtros, `groupby`, archivos) → **Matplotlib / Seaborn** (gráficos que comunican). Con estas cuatro piezas ya puedes cargar un dataset, explorarlo, resumirlo y visualizarlo.

```python
corr = df.drop(columns="species").corr()
sns.heatmap(corr, annot=True, cmap="viridis")
plt.show()
```
