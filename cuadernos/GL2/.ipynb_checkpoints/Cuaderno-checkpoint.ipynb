## 1. NumPy - arreglo n-dimensional

### 1.1. Crear arreglos

```python
a =  np.array([2,4,6,8])
a
```

'np.array()' convierte una lista de Python en un arreglo de NumPy. Al escribir 'a' sola en la última linea, Jupiter muestra su contenido.

```python
np.arange(0, 10, 2)
```

'np.arange(inicio, fin, paso)' genera valores desde 'inicio' hasta 'fin', saltando de 'paso' en 'paso' 

```python
np.linspace(0, 1, 5)
```

'np.linspace(inicio, fin, n)' reparte 'n' valores igualmete expaciados entre 'inicio' y 'fin', ambos incluidos.

'np.zeros(n)', 'np.ones(n)' y 'np.random.rand(n)' tambien crean arreglos: ceros, unos y valores aleatorios.

### 1.2. Forma, tipo y atributos
Todo arreglo conoce su forma 'shape', su tipo de datos 'dtype' y su número de dimensiones 'ndim'. Un arreglo 2D es una matriz de filas y columnas.

Creamos una matriz 2D (dos filas, tres columnas) y consultaremos tres atributos: 'shape' forma, 'dtype' tipo de dato y 'ndim' número de dimensiones.

```python
M = np.array([[1,2,3],
             [4,5,6]])
M.shape, M.dtype, M.ndim
```

### 1.3. Indexado y slicing 
Se accede a los elementos por su posición (empezando de **0**). En 2D, el primer índice es la fila y el segundo la columna; los dos puntos ':' seleccionan todo un eje. 

Con un solo índice seleccionamos una fila completa; 'M[0]' decuelve la primera (los indices empiezan en 0)

```python
M[0]
```

El ':' toma todas las filas y el '1' fija la columna, así obtenemos sa segunda columna.

```python
M[:,1]
```

### 1.4. Operaciones vectorizadas y agregación
Las operaciones se aplican **elemento a elemento** sobre todo el arreglo, sin escribir bucles. Las funciones de agregación resumen los datos, y con 'axis' se agregan por filas o por columnas.

La multiplicación se aplica a cada elemento del arreglo a la vez, sin bucles: esto es una operación vectorizada.

```python
a * 10
```

Los métodos de agregación que resumen el arreglo en un solo número: media, suma y desviación estándar. 

```python
a.mean(), a.sum(), a.std()
```

'axis=0' agrega recorriendo las filas, es decir, suma cada columna por separado.

```python
M.sum(axis=0)
```

### 1.5. Del arreglo al dataset real: iris
A partir de aquí trabajaremos con **iris**, un dataset clásico de 150 flores descritas por 4 medidas (larho y el ancho de sépalo y de pétalo). Viene incluido en scikit-learn y sus datos son un arreglo de NumPy.

Cargar el dataset iris.

```python
from sklearn.datasets import load_iris
iris = load_iris()
iris.data.shape
```

Calculamos la media de cada una de las 4 columnas (características) del arreglo de iris.

```python
iris.data.mean(axis=0)
```

Aquí 'iris.data' es un 'ndarray' puro: útil para el cálculo, pero sin nombres de columnas ni etiquetas legibles. Ese es justamente el trabajo de **Pandas** que veremos a continuación.

## 2. Pandas - series
Una serie es un arreglo **unidimensional etiquetado**: como una columna de una tabla, donde cada valor tiene asociada una etiqueta (su indice). Combina la velocidad de NumPy con un índice que da significado a cada dato.

### 2.1. Crear series

Un conjunto de valores donde cada uno lleva una etiqueta (su índice).

```python
s = pd.Series([5.1, 4.9, 4.7, 4.6],
             index=["m1", "m2", "m3", "m4"])
s
```

### 2.2. Acceso, filtrado y métodos 
Se accede a los valores por etiqueta o por posición, se filtran con condiciones booleanas y se resumen con metodos estadísticos.

Hay dos formas de acceder al mismo valor: por su etiqueta y por su posición.

```python
s["m1"], s.iloc[0]
```

Filtrado booleano: la condicion deja pasar solo los valores que la cumplen.

```python
s[s > 4.7]
```

Métodos estadísticos de las 'Series': promedio, valor máximo

```python
s.mean(), s.max()
```

## 3. Pandas - DataFrames
Un 'Dataframe' es una **tabla bidimensional**, un conjunto de columnas (cada una es una 'Series') que comparten el mismo índice de filas. Es la estructura que más usartemos, representa un data set completo, con sus filas (muestras) y columnas (características).

### 3.1. Crear un DataFrame

Construimos un 'DataFrame' (tabla) a partir de un diccionario: acada clave se convierte en una columna y las listas en sus valores. 

```python
pd.DataFrame({
    "nombre": ["Ana", "Beto", "Carla"],
    "nota": [89, 74, 95]
})
```

### 3.2. Cargar el dataset iris
Construimos un DataFrame con los datos de iris. Usamos nombres de columna cortos y agregamos la especie como clumna **categórica**. 'head()' muestra las primeras 5 filas.

```python
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data,
                 columns = ["sepal_len", "sepal_wid", "petal_len", "petal_wid"])
df["species"] = pd.Categorical.from_codes(iris.target, iris.target_names)
df.head()
```

### 3.2. Explorar el DataFrame
Antes de analizar, conviene mirar la forma y un resumen de los datos. 'shape' devuelve el par (filas, columnas) del DataFrame.

```python
df.shape
```

'describe()' resume una columna numérica: conteo, media, desviación, mínimo, maximo y los cuartiles.

```python
df["sepal_len"].describe()
```

### 3.3. Seleccionar y filtrar
Se selecciona una columna con corchetes y se filtran filas con una condición booleana, igual que NumPy pero conservando etiquetas.

```python
df[df["species"] == "setosa"].head(3)
```

### 3.4. Columnas derivadad y groupby
Podemos crear columnas nuevas a partir de las existentes, y **agrupar** por una categoria para resumir cada grupo. 'groupby' es una de las operaciones más potentes de Pandas.

```python
df["petal_ratio"] = df["petal_len"] / df["petal_wid"]
df[["petal_len", "petal_wid", "petal_ratio"]].head(3)
```

```python
df.groupby("species")[["sepal_len", "petal_len"]].mean()
```

### 3.5. Cargar datos desde un archivo

```python
df.to_csv("iris.csv", index=False)
datos = pd.read_csv("iris.csv")
datos.head(2)
```
